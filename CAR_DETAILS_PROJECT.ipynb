```python
import gradio as gr
import requests

CAR_API_KEY = "14yWr9LYf5Z/KCrlfO4FRw==ePc3DLgZkD3E4JyN"

def fetch_car_details(car_model):
    if not car_model:
        return "❌ Please enter a car model!", None

    url = f"https://api.api-ninjas.com/v1/cars?model={car_model}"
    headers = {"X-Api-Key": CAR_API_KEY}
    response = requests.get(url, headers=headers)

    if response.status_code != 200:
        return "❌ Failed to fetch car data. Check model name or API key.", None

    data = response.json()
    if not data:
        return "❌ Car model not found!", None

    # Use the first result
    car = data[0]
    make = car.get("make", "N/A")
    model = car.get("model", "N/A")
    year = car.get("year", "N/A")
    cylinders = car.get("cylinders", "N/A")
    fuel_type = car.get("fuel_type", "N/A")
    drive = car.get("drive", "N/A")

    # Fake price range logic (since API does not provide price)
    price_range = "₹30-50 Lakhs" if "BMW" in make.upper() else "₹10-25 Lakhs"

    details = f"""
    🚗 **{make} {model}**
    📅 **Year:** {year}
    🔧 **Cylinders:** {cylinders}
    ⛽ **Fuel Type:** {fuel_type}
    🚙 **Drive Type:** {drive}
    💰 **Estimated Price Range:** {price_range}
    """

    return details, None  # API does not return image

# Gradio Interface
with gr.Blocks(css="body {background-color: #121212; color: white;}") as demo:
    gr.Markdown("## 🚗 Car Finder App")

    with gr.Row():
        car_input = gr.Textbox(label="Enter Car Model", placeholder="e.g. BMW")
        search_button = gr.Button("🔍 Search")

    with gr.Row():
        output_text = gr.Markdown()
        output_image = gr.Image(type="pil", label="Car Image")

    search_button.click(fetch_car_details, inputs=car_input, outputs=[output_text, output_image])

demo.launch(share=True, debug=False)
```
